We can use the seaborn library to load a dataset and visualize it. In this case, we will load the 'penguins' dataset and display the first few rows to understand its structure.

```python
import seaborn as sns

pd_data = sns.load_dataset('penguins')
pd_data.head()
```

Now we can use seaborn to create a pairplot of the penguins dataset, which will help us visualize the relationships between different features in the dataset.

```python
pd_data.plot(x='bill_length_mm', y='bill_depth_mm', kind='scatter', s=20, alpha=0.5, color='blue')
```

We can compare the mean and median of the bill length and bill depth features in the penguins dataset to understand their central tendency.

$$ 
\text{mean} = \frac{1}{n}\sum_{i=1}^{n} x_i
$$

and the 

$$
\text{median} = \begin{cases}
x_{(n+1)/2} & \text{if } n \text{ is odd} \\
\frac{x_{n/2} + x_{n/2 + 1}}{2} & \text{if } n \text{ is even}
\end{cases}
$$

```python
mean_length = pd_data['bill_length_mm'].mean()
median_length = pd_data['bill_length_mm'].median()

print(f"Mean of bill_length_mm: {mean_length}")
print(f"Median of bill_length_mm: {median_length}")

mean_depth = pd_data['bill_depth_mm'].mean()
median_depth = pd_data['bill_depth_mm'].median()

print(f"Mean of bill_depth_mm: {mean_depth}")
print(f"Median of bill_depth_mm: {median_depth}")
```

What we can see is that the mean and median of the bill length are quite close, indicating a relatively symmetric distribution.
